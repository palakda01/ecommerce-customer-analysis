```python
import pandas as pd
from getpass import getpass
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

password = getpass("retail_user password: ")

url = URL.create(
    drivername="mysql+pymysql",
    username="retail_user",
    password=password,
    host="localhost",
    port=3306,
    database="retail_analysis"
)
engine = create_engine(url)

with engine.connect() as conn:
    print("Connected successfully")
```

```python
df = pd.read_csv(
    '../data/cleaned/online_retail_cleaned.csv',
    parse_dates=['invoice_date', 'order_date'],
    dtype={'customer_id': 'Int64', 'invoice_no': str, 'stock_code': str}
)
print(df.shape)
print(df.dtypes)
```

```python
df.to_sql(
    'retail_sales',
    engine,
    if_exists='append',
    index=False,
    chunksize=5000,
    method='multi'
)
print("Data loaded")
```
